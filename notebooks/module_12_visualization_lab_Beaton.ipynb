## Part 0: Imports and Dataset Load

## Section

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Palmer Penguins dataset from Seaborn
penguins = sns.load_dataset("penguins")

# Preview the first few rows
print(penguins.head())
```

## Part 1: Inspect the Dataset

## TODO: Display the shape of the dataset

```python
# TODO: Display dataset column information
penguins.info()

# TODO: Display summary statistics for numeric columns
print(penguins.describe())

# TODO: Check for missing values in each column
print(penguins.isna().sum())
```

```python
numeric_cols = penguins.select_dtypes(include='number').columns
categorical_cols = penguins.select_dtypes(exclude='number').columns

print('Numeric columns:', list(numeric_cols))
print('Categorical columns:', list(categorical_cols))
```

## Part 2: Handle Missing Values

## TODO: Create a cleaned version of the dataset by dropping rows with missing values

## Part 3: One Numeric Variable — Histogram

## Task: Create a histogram showing the distribution of penguin body mass

```python
# TODO: Complete the histogram
plt.figure(figsize=(10, 6))

plt.hist(penguins_clean["body_mass_g"], bins=20, edgecolor="black")

plt.title("Distribution of pengiun weight")
plt.xlabel("Weight")
plt.ylabel("Number of Penginus")

plt.show()
```

Reflection Question 3:
What does the histogram tell you about the distribution of penguin body mass?

## Part 4: One Categorical Variable — Count Plot

## Task: Create a count plot showing how many penguins belong to each species.

```python
# TODO: Complete the count plot
sns.countplot(data=penguins_clean, x="species")
plt.xlabel("Species")
plt.ylabel("Count")
plt.title("Number of Penguins by Species")
```

Reflection Question 4:

## Part 5: Numeric vs. Numeric — Scatterplot

## Task: Create a scatterplot showing the relationship between flipper length and body mass

```python
# TODO: Complete the scatterplot
sns.scatterplot(data=penguins_clean, x="flipper_length_mm", y="body_mass_g")
plt.xlabel("Flipper Length (mm)")
plt.ylabel("Body Mass (g)")
plt.title("Flipper Length vs Body Mass")

plt.show()
```

## Part 6: Numeric vs. Categorical — Boxplot

## Task: Create a boxplot comparing body mass across species

```python
# TODO: Complete the boxplot
sns.boxplot(
    data=penguins_clean,
    x="species",
    y="body_mass_g"
)

# Add labels and title
plt.xlabel("Species")
plt.ylabel("Body Mass (g)")
plt.title("Body Mass by Penguin Species")

plt.show()
```

Reflection Question 6:
How does body mass appear to differ by species?
### Overall, body mass is not evenly distributed across species—each species tends to occupy its own weight range, with limited overlap between the smallest and largest groups.

## Part 7: Multiple Variables with hue

## Task: Create a scatterplot showing the relationship between flipper length and body mass, colored by species.

```python
# TODO: Complete the scatterplot with hue
sns.scatterplot(
    data=penguins_clean,
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species"
)

# Add labels and title
plt.xlabel("Flipper Length (mm)")
plt.ylabel("Body Mass (g)")
plt.title("Flipper Length vs Body Mass by Species")

plt.show()
```

Reflection Question 7:
What does adding hue="species" help you see?

## Part 8: Subplots

## Task: Create three histograms (on separate plots) side by side showing the distributions of three numeric variables. 

```python
# TODO: Complete the subplot figure
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Histogram 1: Body Mass
axes[0].hist(penguins_clean["body_mass_g"], bins=20, edgecolor="black")
axes[0].set_title("Body Mass")
axes[0].set_xlabel("grams")
axes[0].set_ylabel("frequency")

# Histogram 2: Flipper Length
axes[1].hist(penguins_clean["flipper_length_mm"], bins=20, edgecolor="black")
axes[1].set_title("Flipper Length")
axes[1].set_xlabel("mm")

# Histogram 3: Bill Length
axes[2].hist(penguins_clean["bill_length_mm"], bins=20, edgecolor="black")
axes[2].set_title("Bill Length")
axes[2].set_xlabel("mm")

plt.tight_layout()
plt.show()
```

Reflection Question 8:
Why might subplots be useful when comparing several related variables?
### patterns are easier to spot

## Part 9: Correlation Heatmap

## Task: Create a correlation heatmap for the numeric columns in the cleaned penguins dataset

```python
# TODO: Fill in the numeric column name
# numeric_cols = penguins_clean.select_dtypes(include='number').columns
numeric_cols = penguins_clean.select_dtypes(include="number")

# TODO: Compute correlation matrix
corr = numeric_cols.corr()

# TODO: Create heatmap
plt.figure(figsize=(6, 5))
sns.heatmap(corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1)

plt.title("Correlation Heatmap (Penguins Dataset)")
plt.show()
```
